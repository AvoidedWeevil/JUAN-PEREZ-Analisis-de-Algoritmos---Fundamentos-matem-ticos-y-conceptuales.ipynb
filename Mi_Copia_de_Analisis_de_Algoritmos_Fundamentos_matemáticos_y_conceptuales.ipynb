**RESPUESTA:**
Funciones piso y techo: Sirven para convertir números con decimales en números enteros exactos cuando medimos recursos de una computadora (como pasos o espacio de memoria). Matemáticamente, si divides un polinomio entre una exponencial cuando los datos son infinitos, el resultado es cero; esto demuestra que la exponencial crece tan rápido que vuelve insignificante a cualquier polinomio, haciendo que el programa sea imposible de resolver en un tiempo razonable.

**RESPUESTA:**
Es un conjunto de instrucciones o pasos claros y ordenados que permiten resolver un problema o calcular un resultado. Sus características clave son que los pasos deben ser sencillos, estar claramente especificados, ejecutarse de manera correcta y lograr el resultado en un tiempo limitado (finito).   
¿Qué es el "Problema del Paro" de Alan Turing?
Consiste en intentar averiguar si cualquier programa de computadora, al recibir ciertos datos, terminará su tarea en algún momento o si, por el contrario, se quedará atrapado repitiendo pasos infinitamente sin detenerse nunca. Turing demostró que este problema no tiene solución (es indecidible); es decir, es imposible crear un programa que pueda predecir esto para todos los casos.   
Diferencia entre Análisis A Priori y Prueba A Posteriori:
Análisis A Priori (Teórico): Es calcular con matemáticas cuántos recursos (tiempo y memoria) va a gastar un algoritmo antes de programarlo. No depende de qué tan rápida sea la computadora ni del lenguaje de programación usado.   
Prueba A Posteriori (Experimental): Es programar el algoritmo y ejecutarlo en una computadora real con datos de prueba para medir con cronómetro cuánto tiempo y memoria gastó después de correrlo. Depende totalmente de la máquina, el lenguaje y los datos específicos que se usaron en esa prueba.

- *Explica en qué consisten el peor caso, mejor caso y caso promedio al momento de evaluar el comportamiento de un algoritmo ante distintas entradas.*  
**RESPUESTA:**  
Peor caso: Es el escenario más difícil posible, donde el algoritmo realiza el máximo de operaciones y tarda más tiempo en terminar. Sirve para garantizar un límite máximo de espera.   
Mejor caso: Es el escenario más fácil o ideal, donde el algoritmo realiza el mínimo de operaciones y termina lo más rápido posible.  
Caso promedio: Es el rendimiento "normal" o esperado en el día a día, calculado promediando el esfuerzo del algoritmo ante diferentes entradas posibles.

Al analizar un algoritmo de ordenamiento sobre una lista, ¿qué escenarios representarían el "caso peor" y el "caso mejor" de los datos de entrada según la literatura revisada?

**RESPUESTA:**
Al analizar un algoritmo de ordenamiento (cuyo objetivo es organizar una secuencia de elementos de menor a mayor), los escenarios de entrada se definen según el esfuerzo requerido:   Caso Peor: Representa la disposición de datos de tamaño $n$ que obliga al algoritmo a realizar la máxima cantidad de operaciones y tardar el mayor tiempo posible. En muchos algoritmos de ordenamiento básicos (como inserción o burbuja), esto ocurre cuando la lista de entrada está completamente ordenada al revés (en orden decreciente), obligando a comparar y mover cada elemento hasta el inicio.   Caso Mejor: Representa la entrada más conveniente de tamaño $n$ para la cual el algoritmo ejecuta el mínimo de operaciones y termina lo más rápido posible. En algoritmos como el ordenamiento por inserción, esto ocurre cuando la lista ya viene previamente ordenada, requiriendo únicamente recorrerla para verificar el orden sin hacer intercambios.

1. Implementa una versión recursiva para el cálculo del factorial ($n! = n \cdot (n-1)!$) y una versión iterativa para Fibonacci.

````python
def factorial_recursivo(n):
    if n == 0 or n == 1: # Caso base
        return 1
    else:
        return n * factorial_recursivo(n - 1) # Llamada recursiva

# Ejemplo de prueba
print("Factorial de 5:", factorial_recursivo(5))
````

````python
def fibonacci_iterativo(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1

    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

# Ejemplo de prueba
print("Fibonacci iterativo de 10:", fibonacci_iterativo(10))
````

2. Utiliza la librería `time` o la función mágica `%timeit` para comparar el tiempo consumido (A posteriori) entre el Fibonacci recursivo y el iterativo para `n=35`.

````python
import time

n = 35

# Medición Fibonacci Recursivo (asegúrate de haber ejecutado la celda 2.1 antes)
inicio_rec = time.time()
res_rec = fibonacci_recursivo(n)
fin_rec = time.time()

# Medición Fibonacci Iterativo
inicio_it = time.time()
res_it = fibonacci_iterativo(n)
fin_it = time.time()

print(f"Fibonacci Recursivo ({n}): {res_rec} -> Tiempo: {fin_rec - inicio_rec:.6f} segundos")
print(f"Fibonacci Iterativo ({n}): {res_it} -> Tiempo: {fin_it - inicio_it:.6f} segundos")
````

3. Elabora un cuadro comparativo concluyendo por qué una de las dos crece de forma polinomial y la otra tiene tintes exponenciales.

| Criterio de Comparación | Fibonacci Iterativo | Fibonacci Recursivo |
| :--- | :--- | :--- |
| **Tipo de Crecimiento** | **Polinomial (Lineal):** $\Theta(n)$ | **Exponencial:** $\Theta(\phi^n)$ / $O(2^n)$ |
| **Estructura de Ejecución** | Utiliza un único ciclo secuencial que avanza paso a paso desde el caso base hasta $n$. | Genera un árbol binario de llamadas donde cada función se divide en dos subproblemas ($n-1$ y $n-2$). |
| **Causa de su Eficiencia / Ineficiencia** | **Sin trabajo repetido:** Calcula cada valor una sola vez y lo reutiliza en la siguiente iteración, ejecutando la operación básica exactamente $n-1$ veces. | **Cálculo redundante:** Recalcula los mismos subproblemas una y otra vez desde cero, multiplicando las operaciones en cada nivel del árbol. |
| **Comportamiento al aumentar $n$** | El tiempo crece de forma proporcional y controlada (si $n$ se duplica, el tiempo solo se duplica). | El tiempo crece de forma explosiva, volviéndose prohibitivamente costoso incluso para valores moderados de $n$. |

**Conclusión:**
La versión **iterativa** crece de forma **polinomial (lineal)** porque resuelve el problema mediante una suma secuencial única donde el número de pasos es directamente proporcional al tamaño de entrada $n$. En cambio, la versión **recursiva** tiene un comportamiento **exponencial** porque cada llamada se ramifica en dos nuevas llamadas sin guardar los resultados previos, provocando que la cantidad de operaciones se multiplique drásticamente con cada incremento de $n$ y supere rápidamente a cualquier función polinomial.
